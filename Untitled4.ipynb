```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------------------------------------------
# 1. CARGA Y PREPARACIÓN DE DATOS DESDE EL ARCHIVO SUBIDO
# -------------------------------------------------------------------
ruta_archivo = 'HUM_REL.csv'

try:
    # Leemos las primeras 30 líneas para identificar con precisión dónde empiezan los datos reales
    with open(ruta_archivo, 'r') as f:
        lineas = f.readlines()[:30]

    skip_rows = 0
    for idx, line in enumerate(lineas):
        # Buscamos la fila de cabecera real que contiene obligatoriamente YEAR y meses
        if 'YEAR' in line and ('JAN' in line or 'Jan' in line):
            skip_rows = idx
            break
        # O si encontramos el indicador estándar de fin de cabecera de la NASA
        elif '-END HEADER-' in line:
            skip_rows = idx + 1
            break

    df = pd.read_csv(ruta_archivo, skiprows=skip_rows)
    print(f"✅ Archivo cargado correctamente saltando {skip_rows} filas de encabezado!")
    print("Columnas encontradas:", df.columns.tolist())
except Exception as e:
    print(f"❌ Error al leer el archivo con autodetección: {e}")
    print("Intentando lectura directa estándar...")
    df = pd.read_csv(ruta_archivo)

# Definición de meses en inglés (como vienen del archivo) y traducción para gráficos
months_en = ['JAN', 'FEB', 'MAR', 'APR', 'MAY', 'JUN', 'JUL', 'AUG', 'SEP', 'OCT', 'NOV', 'DEC']
months_es = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']

# Asegurar que los nombres de las columnas estén en mayúsculas para coincidir
df.columns = df.columns.str.upper()

# Verificar que las columnas de meses existan en el archivo cargado
columnas_faltantes = [m for m in months_en if m not in df.columns]
if columnas_faltantes:
    print(f"⚠️ Advertencia: No se encontraron las columnas {columnas_faltantes} en tu archivo.")

# -------------------------------------------------------------------
# 2. CÁLCULO DE LA MATRIZ MULTIANUAL
# -------------------------------------------------------------------
multianual = pd.DataFrame({
    'Mes': months_es,
    'Promedio': df[months_en].mean().values,
    'Minimo': df[months_en].min().values,
    'Maximo': df[months_en].max().values
})

print("\n=== PROMEDIOS MULTIANUALES ===")
print(multianual.round(2))
print(f"Promedio Anual Multianual: {multianual['Promedio'].mean():.2f}%")

# -------------------------------------------------------------------
# 3. TRANSFORMACIÓN A SERIE ESTACIONAL CONTINUA (AÑO-MES)
# -------------------------------------------------------------------
df_long = df.melt(id_vars=['YEAR'], value_vars=months_en, var_name='MONTH_NAME', value_name='VALOR')
# Mapa de meses a números
month_map = {m: i+1 for i, m in enumerate(months_en)}
df_long['MONTH'] = df_long['MONTH_NAME'].map(month_map)
df_long['FECHA'] = pd.to_datetime(df_long['YEAR'].astype(str) + '-' + df_long['MONTH'].astype(str) + '-01')
df_long = df_long.sort_values('FECHA').reset_index(drop=True)

# -------------------------------------------------------------------
# 4. GENERACIÓN DE GRÁFICOS
# -------------------------------------------------------------------
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, axes = plt.subplots(2, 1, figsize=(12, 9))

# GRÁFICO 1: Serie Temporal Estacional (Historico continuo)
axes[0].plot(df_long['FECHA'], df_long['VALOR'], color='#1B365D', linewidth=1.8, label='Humedad Relativa (%)')
axes[0].axhline(df_long['VALOR'].mean(), color='#D9534F', linestyle='--', linewidth=1.2, label=f'Promedio Histórico ({df_long["VALOR"].mean():.1f}%)')
axes[0].set_title('1. SERIE TEMPORAL ESTACIONAL (HISTÓRICO AÑO A AÑO)', fontsize=12, fontweight='bold', color='#1B365D')
axes[0].set_ylabel('Humedad Relativa (%)', fontsize=10)
axes[0].set_xlabel('Año', fontsize=10)
axes[0].legend(loc='upper right')
axes[0].grid(True, alpha=0.4)

# GRÁFICO 2: Ciclo Multianual Promedio (12 Meses)
axes[1].plot(multianual['Mes'], multianual['Promedio'], color='#2C5E8A', marker='o', linewidth=2.5, label='Promedio Multianual')
axes[1].fill_between(multianual['Mes'], multianual['Minimo'], multianual['Maximo'], color='#2C5E8A', alpha=0.15, label='Rango de Variabilidad (Min-Max)')
axes[1].set_title('2. CICLO CLIMATOLÓGICO MULTIANUAL (PROMEDIO MENSUAL)', fontsize=12, fontweight='bold', color='#1B365D')
axes[1].set_ylabel('Humedad Relativa (%)', fontsize=10)
axes[1].set_xlabel('Mes del Año', fontsize=10)
axes[1].legend(loc='upper right')
axes[1].grid(True, alpha=0.4)

plt.tight_layout()
plt.savefig('Caracterizacion_Humedad_Relativa.png', dpi=300)
plt.show()
```
